```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries loaded successfully!")
```

```python
df = pd.read_csv("global_ecommerce_sales.csv")

df.head()
```

```python
df.columns
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()

print("Duplicate rows:", df.duplicated().sum())
```

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df.info()
```

```python
print("First Order Date:", df["Order_Date"].min())
print("Last Order Date:", df["Order_Date"].max())
```

```python
df[["Quantity", "Unit_Price", "Discount_Percent", "Total_Sales", "Shipping_Cost", "Profit"]].describe()
```

```python
total_sales = df["Total_Sales"].sum()

print("Total Sales:", total_sales)
```

```python
total_profit = df["Profit"].sum()

print("Total Profit:", total_profit)
```

```python
total_orders = df["Order_ID"].nunique()

print("Total Orders:", total_orders)
```

```python
total_quantity = df["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
average_order_value = df["Total_Sales"].mean()

print("Average Order Value:", average_order_value)
```

```python
category_sales = df.groupby("Product_Category")["Total_Sales"].sum().sort_values(ascending=False)

print(category_sales)
```

```python
category_profit = df.groupby("Product_Category")["Profit"].sum().sort_values(ascending=False)

print(category_profit)
```

```python
plt.figure(figsize=(8,5))

category_sales.plot(kind="bar")

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

category_profit.plot(kind="bar")

plt.title("Profit by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
region_sales = df.groupby("Region")["Total_Sales"].sum().sort_values(ascending=False)

print(region_sales)
```

```python
region_profit = df.groupby("Region")["Profit"].sum().sort_values(ascending=False)

print(region_profit)
```

```python
plt.figure(figsize=(8,5))

region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

region_profit.plot(kind="bar")

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df["Month"] = df["Order_Date"].dt.to_period("M")

df[["Order_Date", "Month"]].head()
```

```python
monthly_sales = df.groupby("Month")["Total_Sales"].sum()

print(monthly_sales)
```

```python
plt.figure(figsize=(12,5))

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
best_month = monthly_sales.idxmax()
best_month_sales = monthly_sales.max()

print("Highest Sales Month:", best_month)
print("Sales:", best_month_sales)
```

```python
top_products = (
    df.groupby("Product_Name")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(top_products)
```

```python
plt.figure(figsize=(10,5))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python
best_product = top_products.index[0]
best_product_sales = top_products.iloc[0]

print("Top Product:", best_product)
print("Sales:", best_product_sales)
```

```python
customer_sales = (
    df.groupby("Customer_Name")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print(customer_sales)
```

```python
plt.figure(figsize=(10,5))

customer_sales.sort_values().plot(kind="barh")

plt.title("Top 10 Customers by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Customer")
plt.tight_layout()
plt.show()
```

```python
top_customer = customer_sales.index[0]
top_customer_sales = customer_sales.iloc[0]

print("Top Customer:", top_customer)
print("Sales:", top_customer_sales)
```

```python
segment_sales = (
    df.groupby("Customer_Segment")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(segment_sales)
```

```python
plt.figure(figsize=(8,5))

segment_sales.plot(kind="bar")

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
segment_profit = (
    df.groupby("Customer_Segment")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print(segment_profit)
```

```python
plt.figure(figsize=(8,5))

segment_profit.plot(kind="bar")

plt.title("Profit by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Profit")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Correlation Analysis

correlation = df[
    ["Quantity", "Unit_Price", "Discount_Percent",
     "Total_Sales", "Shipping_Cost", "Profit"]
].corr()

print(correlation)
```
